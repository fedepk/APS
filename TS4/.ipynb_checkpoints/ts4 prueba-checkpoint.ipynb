## Introducción
La Transformada Rápida de Fourier (FFT) es una herramienta que se utiliza, entre otras aplicaciones, para calcular el espectro de potencia de una señal determinada. Esta operación presenta un fenómeno particular, conocido como *desparramo espectral* o *spectral leakage*, que ocurre cuando se analizan frecuencias cuyo período no entra un número entero de veces dentro de la ventana de muestreo. 

En este trabajo realizaremos la experiencia de calcular las densidades espectrales de potencia evaluando la estimación de amplitud y frecuencia de una señal senoidal contaminada con ruido, para distintas Relaciones Señal a Ruido (SNR). El experimento se repetirá $M=200$ veces (Monte Carlo) considerando una desintonía aleatoria en la frecuencia de la señal, y se analizará el comportamiento de diferentes ventanas: Rectangular, Flat-top, Blackman-Harris y Hann.

---

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal.windows as sp

#%% --------------------------------- 1. Definiciones

fs = 1000
N = 1000
a = 2
v_max = np.sqrt(2)
M = 200
snrs = [3, 10]

# Frecuencia central nominal
f0 = fs / 4               # 250 Hz
bin_f0 = N // 4           # Bin 250 (Omega_0 = pi/2)
omega_0 = 2 * np.pi * f0 / fs 

# Vector de frecuencias para el ploteo del espectro (Nyquist)
frec = np.arange(N // 2) * (fs / N)

#%% --------------------------------- 2. Definición de Ventanas

ventanas = {
    'Rectangular': np.ones(N),
    'Flat-top': sp.flattop(N),
    'Blackman': sp.blackmanharris(N),
    'Hann': sp.hann(N)
}

# Diccionarios para guardar resultados
results = {snr: {name: {} for name in ventanas} for snr in snrs}

#%% --------------------------------- 3. Simulación Monte Carlo (Vectorizada)

np.random.seed(42)
fr = np.random.uniform(-a, a, M)
f = (fs / 4 + fr * fs / N)                
omega_1 = f * (2 * np.pi / fs)            

tt = np.arange(0, N, 1).reshape(N, 1)

for snr in snrs:
    
    xx = v_max * np.sin(2 * np.pi * f * tt / fs)
    pot = np.sqrt(10**(-snr/10))
    n_a = np.random.normal(0, pot, (N, M))
    xx = xx + n_a
    
    for nombre_vent, ventana in ventanas.items():
        ventana_col = ventana.reshape(N, 1)
        x_v = xx * ventana_col
        
        # FFT a lo largo de las columnas
        X_V = np.fft.fft(x_v, axis=0)
        X_V_mitad = X_V[:(N//2), :]
        
        # --- 1. Estimadores de Amplitud y Frecuencia ---
        escala_amp = 2 / np.sum(ventana)
        a_est = np.abs(X_V_mitad[bin_f0, :]) * escala_amp
        
        bines_pico = np.argmax(np.abs(X_V_mitad), axis=0)
        omega_est = bines_pico * (2 * np.pi / N)
        
        # --- 2. Cálculo de la Densidad Espectral de Potencia (PSD) ---
        S2 = np.sum(ventana**2)
        psd = (2 / (fs * S2)) * (np.abs(X_V_mitad[:, 0])**2)
        psd_db = 10 * np.log10(psd)
        
        # Guardamos datos
        results[snr][nombre_vent]['a_est'] = a_est
        results[snr][nombre_vent]['omega_est'] = omega_est
        results[snr][nombre_vent]['psd_db'] = psd_db

#%% --------------------------------- 4. Ploteo de Histogramas

for snr in snrs:
    fig, ax = plt.subplots(figsize=(10, 6), layout="constrained")
    ax.set_title(f'Distribución de Estimación de Amplitud (SNR = {snr} dB)', fontsize=12, fontweight='bold')
    ax.set_xlabel(r'Amplitud estimada $\hat{a}_1$')
    ax.set_ylabel('Frecuencia Relativa')
    
    for nombre_vent in ventanas.keys():
        ax.hist(results[snr][nombre_vent]['a_est'], bins=20, alpha=0.5, label=nombre_vent)
        
    ax.axvline(v_max, color='red', linestyle='dashed', linewidth=2, label=f'Valor Real ($a_0={v_max:.3f}$)')
    ax.legend()
    ax.grid(True, linestyle=':', alpha=0.7)
    plt.show()

#%% --------------------------------- 5. Ploteo de PSD

for snr in snrs:
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5), layout="constrained")
    fig.suptitle(f"Densidad Espectral de Potencia (SNR = {snr} dB) - 1 Realización", fontsize=14, fontweight='bold')
    
    # Vista general
    for nombre_vent in ventanas.keys():
        ax1.plot(frec, results[snr][nombre_vent]['psd_db'], label=nombre_vent, linewidth=1)
    ax1.set_xlim([0, fs/2])
    ax1.set_title("Espectro Completo")
    ax1.set_ylabel("PSD [dB/Hz]")
    ax1.set_xlabel("Frecuencia [Hz]")
    ax1.grid(True, which='both', linestyle=':', alpha=0.5)
    ax1.minorticks_on()
    
    # Zoom
    for nombre_vent in ventanas.keys():
        ax2.plot(frec, results[snr][nombre_vent]['psd_db'], label=nombre_vent, linewidth=1.5)
    ax2.set_xlim([240, 260])
    ax2.set_ylim([-50, 5])
    ax2.set_title("Zoom al lóbulo principal")
    ax2.set_xlabel("Frecuencia [Hz]")
    ax2.grid(True, which='both', linestyle=':', alpha=0.5)
    ax2.minorticks_on()
    ax2.legend()
    
    plt.show()

#%% --------------------------------- 6. Tablas de Sesgo y Varianza

def imprimir_tablas():
    for snr in snrs:
        print(f"\n{'='*55}")
        print(f"   RESULTADOS PARA SNR = {snr} dB")
        print(f"{'='*55}")
        print("\n--- Estimación de Amplitud ---")
        print(f"{'Ventana':<15} | {'Sesgo (s_a)':<15} | {'Varianza (v_a)':<15}")
        print("-" * 50)
        
        for nombre_vent in ventanas.keys():
            a_est = results[snr][nombre_vent]['a_est']
            s_a = np.mean(a_est) - v_max
            v_a = np.var(a_est)
            print(f"{nombre_vent:<15} | {s_a:>15.6f} | {v_a:>15.6f}")

        print("\n--- Estimación de Frecuencia ---")
        print(f"{'Ventana':<15} | {'Sesgo (s_f)':<15} | {'Varianza (v_f)':<15}")
        print("-" * 50)
        
        for nombre_vent in ventanas.keys():
            omega_est = results[snr][nombre_vent]['omega_est']
            error_f = omega_est - omega_1
            s_f = np.mean(error_f)
            v_f = np.var(error_f)
            print(f"{nombre_vent:<15} | {s_f:>15.6f} | {v_f:>15.6f}")
        print("\n")

imprimir_tablas()
```

## 3. Conclusiones y Fenómenos Observados

A partir de los histogramas, densidades espectrales y las tablas de varianza y sesgo generadas, podemos extraer las siguientes conclusiones sobre la estimación espectral:

### El efecto *Scalloping Loss* (Pérdida por festoneo)
En los histogramas de amplitud se evidencia que la ventana **Rectangular** presenta un fuerte sesgo negativo. Esto ocurre porque su lóbulo principal es sumamente estrecho en el dominio frecuencial (como se ve en los gráficos PSD). Dado que la frecuencia real de la señal posee una desintonía aleatoria, muchas veces no coincide exactamente con el *bin* de la DFT evaluado ($k_0 = \frac{N}{4}$). Al evaluar rígidamente en ese bin, la ventana rectangular atenúa fuertemente la señal si el pico cayó "entre medio" de dos bines, provocando que la amplitud estimada sea sistemáticamente menor a la real.

### El compromiso Sesgo vs. Varianza (El rol de la ventana Flat-top)
La ventana **Flat-top** fue diseñada específicamente para mitigar el *scalloping loss*. Como se observa en la vista magnificada de la PSD, su lóbulo principal es intencionalmente ancho y achatado en la cima. Esto garantiza que, aunque la frecuencia fluctúe $\pm 2$ bines por la desintonía, la ganancia aplicada sobre el bin central permanezca prácticamente constante. Como resultado:
*   **Sesgo casi nulo:** El histograma de la Flat-top se centra perfectamente sobre el valor real de amplitud ($a_0 = \sqrt{2}$).
*   **Varianza elevada:** El costo de tener un lóbulo principal "gordo" es que la ventana presenta un **Ancho de Banda Equivalente de Ruido (ENBW)** muy alto. Esto significa que deja pasar mucha más potencia del piso de ruido ($n_a$) hacia la medición del bin, empeorando dramáticamente la varianza de la estimación de amplitud (fenómeno muy visible cuando la SNR es de 3 dB).

### Ventanas de compromiso: Blackman-Harris y Hann
Estas ventanas representan un punto medio. Sus lóbulos principales son más anchos que la rectangular (mitigando parcialmente el *scalloping loss* y mejorando el sesgo) pero más finos que la Flat-top (limitando el paso del piso de ruido y mejorando la varianza respecto a esta). Además, como se observa en la PSD general, poseen lóbulos laterales con excelente atenuación, evitando el *spectral leakage* desde frecuencias alejadas.

### El efecto del Zero-Padding
Si en este experimento aplicáramos *zero-padding* agregando ceros al final de cada secuencia antes de aplicar la FFT, observaríamos dos fenómenos principales:
1.  **Interpolación Espectral:** Las densidades espectrales se verían mucho más suaves, permitiendo apreciar la forma continua de la función `sinc` y sus lóbulos.
2.  **Mejora en la Estimación de Frecuencia:** Al no aportar nueva información, el zero-padding no mejora la resolución intrínseca (no permite separar tonos más cercanos). Sin embargo, al densificar la cuadrícula de puntos evaluados en el espectro (menor $\Delta f$), el estimador del máximo espectral ($\hat{\Omega}_1 = \arg\max$) tiene muchas más opciones de bines donde "caer", reduciendo masivamente el error de cuantización por discretización, lo que resulta en una caída drástica de la **varianza de frecuencia ($v_f$)**.
