## Experiencia 1: Señal Sinusoidal de 2 kHz (Condición de Muestreo y Fuga Espectral)

### Marco Teórico
La consigna exige que la señal senoidal posea **al menos 10 puntos por período** ($N_{per} \ge 10$):
$$N_{per} = \frac{F_s}{f_0} \ge 10 \implies F_s \ge 10 \cdot 2000\text{ Hz} = 20\text{ kHz}$$

La resolución espectral de la DFT está fijada por $\Delta f = \frac{F_s}{N}$. El número de bin discreto en el que cae una señal es:
$$k_0 = \frac{f_0}{\Delta f} = \frac{f_0 \cdot N}{F_s}$$

Se comparan dos frecuencias de muestreo:
1. **$F_s = 25\text{ kHz}$:** $k_0 = \frac{2000 \cdot 1000}{25000} = 80 \in \mathbb{Z}$. La ventana temporal contiene exactamente 80 ciclos completos. El espectro resultante es ortogonal a los bines de la FFT y no presenta fuga.
2. **$F_s = 22\text{ kHz}$:** $k_0 = \frac{2000 \cdot 1000}{22000} \approx 90.909 \notin \mathbb{Z}$. La ventana contiene 90.91 ciclos, provocando una discontinuidad artificial en los bordes que dispersa energía hacia frecuencias adyacentes (*spectral leakage*).

```python
# Caso 1: Fs = 25 kHz (Sin fuga espectral)
Fs_1 = 25000
t_1 = np.arange(N) / Fs_1
x_1 = np.sin(2 * np.pi * f0 * t_1)

X_1 = (1 / N) * np.fft.fft(x_1)
frec_1 = np.fft.fftfreq(N, 1 / Fs_1)[:N // 2]
X_1_db = 10 * np.log10(2 * (np.abs(X_1[:N // 2]) ** 2) + 1e-12)

# Caso 2: Fs = 22 kHz (Con fuga espectral)
Fs_2 = 22000
t_2 = np.arange(N) / Fs_2
x_2 = np.sin(2 * np.pi * f0 * t_2)

X_2 = (1 / N) * np.fft.fft(x_2)
frec_2 = np.fft.fftfreq(N, 1 / Fs_2)[:N // 2]
X_2_db = 10 * np.log10(2 * (np.abs(X_2[:N // 2]) ** 2) + 1e-12)

# Graficación comparativa
fig, ax = plt.subplots(2, 2, figsize=(13, 7), layout="constrained")

# Señales temporales (primeras 100 muestras)
ax[0, 0].plot(t_1[:100] * 1e3, x_1[:100], '.-b')
ax[0, 0].set_title(r"Señal en el Tiempo ($F_s = 25\text{ kHz}$, 12.5 pts/ciclo)")
ax[0, 0].set_xlabel("Tiempo [ms]")
ax[0, 0].set_ylabel("Amplitud [V]")

ax[0, 1].plot(t_2[:100] * 1e3, x_2[:100], '.-r')
ax[0, 1].set_title(r"Señal en el Tiempo ($F_s = 22\text{ kHz}$, 11 pts/ciclo)")
ax[0, 1].set_xlabel("Tiempo [ms]")
ax[0, 1].set_ylabel("Amplitud [V]")

# Módulos de FFT
ax[1, 0].stem(frec_1, X_1_db, basefmt=" ")
ax[1, 0].set_title("Espectro Puro (Sin fuga, energía confinada en bin 80)")
ax[1, 0].set_xlabel("Frecuencia [Hz]")
ax[1, 0].set_ylabel("Potencia [dB]")
ax[1, 0].set_xlim(0, 5000)

ax[1, 1].plot(frec_2, X_2_db, 'r.-')
ax[1, 1].set_title("Espectro con Fuga (Discontinuidad perimetral en la ventana)")
ax[1, 1].set_xlabel("Frecuencia [Hz]")
ax[1, 1].set_ylabel("Potencia [dB]")
ax[1, 1].set_xlim(0, 5000)

plt.show()
```

#### Conclusiones de la Experiencia 1:
1. **Cumplimiento de Nyquist:** En ambos casos ($F_s = 25\text{ kHz}$ y $F_s = 22\text{ kHz}$) se satisface holgadamente el criterio de Nyquist ($F_s > 2 f_0 = 4\text{ kHz}$) y la restricción de poseer al menos 10 puntos por período.
2. **Origen de la anomalía con 22 kHz:** La FFT asume implícitamente que el registro de $N$ muestras se repite periódicamente en el tiempo de $-\infty$ a $+\infty$. Al muestrear a 22 kHz, el bloque temporal termina en $90.91$ ciclos; la unión entre el final y el principio de la ventana genera un salto brusco (discontinuidad artificial), ensuciando el espectro con componentes armónicas de alta frecuencia que no pertenecen a la señal real.
3. **Efecto de Fuga Espectral (*Spectral Leakage*):** En 25 kHz, la componente cae exactamente en el bin frecuencial $k=80$ y toda la potencia queda concentrada en una única línea. En 22 kHz, la frecuencia cae en $k \approx 90.91$ (entre los bines 90 y 91), reduciendo la amplitud del pico principal y ensanchando su base en forma de lóbulos laterales.

## Experiencia 2: Señal de Potencia Media 2 W y Desfasada en $\pi/2$

### Marco Teórico
La potencia media de una señal periódica pura de forma $x(t) = A \sin(2\pi f_0 t + \phi)$ es:
$$P = \frac{A^2}{2} \implies A = \sqrt{2P} = \sqrt{2 \cdot 2\text{ W}} = 2\text{ V}$$

Un adelanto de fase $\phi = \pi/2$ transforma la función seno en una componente cosenoidal pura:
$$x(t) = 2 \sin(2\pi f_0 t + \pi/2) = 2 \cos(2\pi f_0 t)$$

```python
Fs = 25000
t = np.arange(N) / Fs
P_deseada = 2.0  # Watts
A = np.sqrt(2 * P_deseada)
phi = np.pi / 2

x_desfasada = A * np.sin(2 * np.pi * f0 * t + phi)

# FFT normalizada
X_desf = (1 / N) * np.fft.fft(x_desfasada)
frec = np.fft.fftfreq(N, 1 / Fs)[:N // 2]
X_desf_mod = np.abs(X_desf[:N // 2])

# Gráficos
fig, ax = plt.subplots(2, 1, figsize=(10, 6), layout="constrained")
ax[0].plot(t[:100] * 1e3, x_desfasada[:100], '.-g')
ax[0].set_title(rf"Señal Senoidal Desfasada: $A = {A:.1f}\text{{ V}}$, $\phi = \pi/2$, $P = {P_deseada}\text{{ W}}$")
ax[0].set_xlabel("Tiempo [ms]")
ax[0].set_ylabel("Amplitud [V]")

ax[1].stem(frec, X_desf_mod, basefmt=" ")
ax[1].set_title("Módulo de la Transformada de Fourier")
ax[1].set_xlabel("Frecuencia [Hz]")
ax[1].set_ylabel("Amplitud Normalizada [V]")
ax[1].set_xlim(0, 5000)
plt.show()
```

#### Conclusiones de la Experiencia 2:
1. **Independencia del módulo frente al desfase:** El módulo de la FFT $|X(f)|$ de la señal desfasada en $\pi/2$ presenta exactamente la misma distribución frecuencial que una senoidal en fase. El desfase sólo altera la parte imaginaria de los coeficientes de Fourier (espectro de fase $\angle X(f)$).
2. **Amplitud en el espectro unilateral y bilateral:** La FFT normalizada por $1/N$ divide la energía de una señal cosenoidal real de amplitud $A=2\text{ V}$ en dos deltas de valor $A/2 = 1\text{ V}$, situadas en $+f_0$ y $-f_0$. Por este motivo, el pico positivo en el gráfico unilateral mide exactamente 1 V.

## Experiencias 3 y 4: Ruido Gaussiano y Ruido Uniforme

### Marco Teórico
Para una variable aleatoria de media nula (sin componente de continua $\mu = 0\text{ V}$), la potencia media equivale a su varianza: $P = \sigma^2 = 0.1\text{ W}$.

* **Ruido Gaussiano:**
  $$X \sim \mathcal{N}(\mu = 0, \sigma^2 = 0.1) \implies \sigma = \sqrt{0.1} \approx 0.3162\text{ V}$$
* **Ruido Uniforme:**
  Definido en el intervalo simétrico $[-a, a]$, su varianza teórica es:
  $$\sigma^2 = \frac{(2a)^2}{12} = \frac{a^2}{3} = 0.1 \implies a = \sqrt{3 \cdot 0.1} = \sqrt{0.3} \approx 0.5477\text{ V}$$

En ambos casos se evalúa la señal temporal, el histograma comparado con su densidad de probabilidad analítica, y el módulo de la FFT.

```python
sigma2 = 0.1

# 1. Ruido Normal (Gaussiano)
desv_gauss = np.sqrt(sigma2)
ruido_norm = np.random.normal(0, desv_gauss, N)

# 2. Ruido Uniforme
lim_unif = np.sqrt(3 * sigma2)
ruido_unif = np.random.uniform(-lim_unif, lim_unif, N)

# FFTs
X_norm = (1 / N) * np.fft.fft(ruido_norm)
X_unif = (1 / N) * np.fft.fft(ruido_unif)

# Subplots integrados
fig, ax = plt.subplots(3, 2, figsize=(13, 9), layout="constrained")

# Dominio temporal
ax[0, 0].plot(ruido_norm, '.', markersize=4, color='royalblue')
ax[0, 0].set_title(rf"Ruido Gaussiano ($\mu=0\text{{ V}}$, $\sigma^2={sigma2}\text{{ W}}$)")
ax[0, 0].set_ylabel("Amplitud [V]")

ax[0, 1].plot(ruido_unif, '.', markersize=4, color='darkorange')
ax[0, 1].set_title(rf"Ruido Uniforme ($\mu=0\text{{ V}}$, $\sigma^2={sigma2}\text{{ W}}$)")
ax[0, 1].set_ylabel("Amplitud [V]")

# Histogramas con curvas de densidad superpuestas
ax[1, 0].hist(ruido_norm, bins=30, density=True, edgecolor='black', alpha=0.6, color='royalblue')
x_axis_g = np.linspace(-3*desv_gauss, 3*desv_gauss, 100)
ax[1, 0].plot(x_axis_g, norm.pdf(x_axis_g, 0, desv_gauss), 'r', lw=2)
ax[1, 0].set_title("Histograma vs FDP Normal Teórica")
ax[1, 0].set_ylabel("Densidad")

ax[1, 1].hist(ruido_unif, bins=30, density=True, edgecolor='black', alpha=0.6, color='darkorange')
x_axis_u = np.linspace(-lim_unif*1.2, lim_unif*1.2, 100)
ax[1, 1].plot(x_axis_u, uniform.pdf(x_axis_u, -lim_unif, 2*lim_unif), 'r', lw=2)
ax[1, 1].set_title("Histograma vs FDP Uniforme Teórica")
ax[1, 1].set_ylabel("Densidad")

# Espectro de magnitud
ax[2, 0].plot(frec, np.abs(X_norm[:N // 2]), color='royalblue')
ax[2, 0].set_title("Módulo de la FFT (Densidad Espectral Plana)")
ax[2, 0].set_xlabel("Frecuencia [Hz]")
ax[2, 0].set_ylabel("|X(f)|")

ax[2, 1].plot(frec, np.abs(X_unif[:N // 2]), color='darkorange')
ax[2, 1].set_title("Módulo de la FFT (Densidad Espectral Plana)")
ax[2, 1].set_xlabel("Frecuencia [Hz]")
ax[2, 1].set_ylabel("|X(f)|")

plt.show()
```

#### Conclusiones de las Experiencias 3 y 4:
1. **Distribución en amplitud vs. distribución en frecuencia:** Aunque las dos secuencias provienen de procesos estocásticos completamente diferentes en el dominio de la amplitud (el gaussiano concentra valores según la campana y tiene colas infinitas, mientras que el uniforme está acotado estrictamente entre $\pm 0.548\text{ V}$), **sus espectros son indistinguibles a simple vista**.
2. **Naturaleza del Ruido Blanco:** Ambas secuencias carecen de correlación temporal entre muestras consecutivas, por lo que su Densidad Espectral de Potencia (PSD) es constante a lo largo de toda la banda de Nyquist ($[0, F_s/2]$). No presentan líneas discretas armónicas, sino un piso plano con fluctuaciones aleatorias alrededor de su valor medio espectral.

## Experiencia 5: Pulso Rectangular Periódico (Onda Cuadrada)

### Marco Teórico
Se sintetiza una onda periódica cuadrada bipolar de ciclo de trabajo del 50% ($D = 0.5$) con valores en $\pm A$.
Su potencia media es constante e independiente del tiempo:
$$P = \frac{1}{T} \int_{0}^{T} x^2(t) dt = A^2 = 1\text{ W} \implies A = \sqrt{1\text{ W}} = 1\text{ V}$$

El desarrollo en serie de Fourier de una onda cuadrada simétrica sólo contiene armónicos impares ($f_0, 3f_0, 5f_0, \dots$), cuya amplitud decae proporcionalmente a $\frac{1}{k}$.

```python
P_pulso = 1.0
A_pulso = np.sqrt(P_pulso)

# Generación mediante scipy.signal.square
pulso = A_pulso * signal.square(2 * np.pi * f0 * t, duty=0.5)

# FFT normalizada
X_pulso = (1 / N) * np.fft.fft(pulso)
X_pulso_mod = np.abs(X_pulso[:N // 2])

fig, ax = plt.subplots(2, 1, figsize=(10, 6), layout="constrained")
ax[0].plot(t[:100] * 1e3, pulso[:100], '.-m')
ax[0].set_title(f"Pulso Rectangular Periódico: f = {f0} Hz, Ciclo de trabajo = 50%, P = {P_pulso} W")
ax[0].set_xlabel("Tiempo [ms]")
ax[0].set_ylabel("Amplitud [V]")

ax[1].stem(frec, X_pulso_mod, basefmt=" ")
ax[1].set_title("Módulo de la FFT (Armónicos impares $f_0, 3f_0, 5f_0$ con decaimiento $1/k$)")
ax[1].set_xlabel("Frecuencia [Hz]")
ax[1].set_ylabel("|X(f)|")
ax[1].set_xlim(0, 10000)
plt.show()
```

#### Conclusiones de la Experiencia 5:
1. **Contenido armónico de la onda cuadrada:** Se evidencia claramente la ausencia de armónicos pares ($2f_0 = 4\text{ kHz}$, $4f_0 = 8\text{ kHz}$), característico de funciones con simetría de media onda impar: $x(t) = -x(t + T/2)$.
2. **Velocidad de convergencia espectral:** Al poseer discontinuidades de salto en el tiempo, los coeficientes de Fourier decaen lentamente a razón de $1/k$ (equivalente a un decaimiento de $-20\text{ dB/década}$). Esto exige un ancho de banda significativamente mayor que el de una senoidal pura para no distorsionar la señal.

## Bonus: Onda Triangular Adicional y Comprobación del Teorema de Parseval

### Marco Teórico del Teorema de Parseval
Para una señal discreta $x[n]$ y su Transformada Discreta de Fourier normalizada $X[k] = \frac{1}{N}\sum_{n=0}^{N-1} x[n] e^{-j\frac{2\pi}{N}kn}$, la potencia media en ambos dominios satisface estrictamente la igualdad:
$$P = \frac{1}{N} \sum_{n=0}^{N-1} |x[n]|^2 = \sum_{k=0}^{N-1} |X[k]|^2$$

Implementamos una señal adicional mediante `scipy.signal.sawtooth(..., width=0.5)` (onda triangular periódica) y tabulamos la verificación numérica en todas las señales desarrolladas en el TP.

```python
# Señal adicional: Onda triangular de 2 kHz
triangular = signal.sawtooth(2 * np.pi * f0 * t, width=0.5)
X_triang = (1 / N) * np.fft.fft(triangular)

# Diccionario con todas las señales analizadas en el trabajo práctico
senales = {
    "Senoidal pura (Fs=25kHz)": x_1,
    "Senoidal (P=2W, phi=pi/2)": x_desfasada,
    "Ruido Gaussiano (sigma2=0.1)": ruido_norm,
    "Ruido Uniforme (sigma2=0.1)": ruido_unif,
    "Pulso Cuadrado (Duty=50%)": pulso,
    "Bonus: Onda Triangular": triangular
}

print("=" * 85)
print(f"{'Señal Sintetizada':<30} | {'P. Tiempo [W]':<16} | {'P. Frecuencia [W]':<16} | {'Error Absoluto':<15}")
print("=" * 85)

for nombre, senal_vec in senales.items():
    # Potencia calculada en el dominio del tiempo
    pot_t = np.mean(senal_vec ** 2)
    
    # Potencia calculada a través del espectro normalizado (Parseval)
    X_k = (1 / len(senal_vec)) * np.fft.fft(senal_vec)
    pot_f = np.sum(np.abs(X_k) ** 2)
    
    error = np.abs(pot_t - pot_f)
    print(f"{nombre:<30} | {pot_t:<16.6f} | {pot_f:<16.6f} | {error:<15.2e}")
print("=" * 85)
```

#### Conclusiones del Teorema de Parseval y Bonus:
1. **Conservación exacta de la energía:** En todas las familias de señales evaluadas (armónicas continuas, secuencias estocásticas y ondas periódicas con transiciones bruscas), el error absoluto entre la potencia media temporal y la frecuencial se ubicó en el orden de $10^{-16} \text{ W}$, lo que demuestra la equivalencia unitaria estricta de la base ortogonal de la DFT.
2. **Onda Triangular vs Onda Cuadrada:** La onda triangular posee armónicos que decaen con $1/k^2$ (a razón de $-40\text{ dB/década}$), mucho más rápido que la onda cuadrada ($-20\text{ dB/década}$). Esto se debe a que la función triangular es continua en todo su recorrido temporal, teniendo discontinuidades recién en su primera derivada.
