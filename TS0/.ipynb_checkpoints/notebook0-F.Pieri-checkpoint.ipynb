# Introducción
Realización de una fucnión en pyton que genera una onda senoidal, la cual se puede modificar mediante parámetros.

```python
# -*- coding: utf-8 -*-
"""
Created on Wed Aug 12 09:03:20 2026

@author: Fede
"""
import matplotlib.pyplot as plt
import numpy as np


#%%  Definiciones


#%%  Funcion generador de señales

def generador( vmax , dc , ff , ph , nn , fs):
    
    tt=np.arange(0,nn,1)

    xx = vmax * np.sin(2*np.pi*ff*tt/fs + ph) + dc
    
    return tt , xx
```

Parámetros:  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;vmax: Valor máximo de amplitud de la senoidal.  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;dc: Valor de offset de la señal.  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;ff: Frecuencia.  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;ph: Fase (Valor entre 0 a 2pi).  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;nn: Cantidad de muestras.  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;fs: Frecuencia de muestreo  
Devuelve:  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;tt,xx: Vector de valores de tiempo (tt) y tensión (xx) 

Ejemplos de funcionamiento:

```python
tt , xx = generador(1,0,2,0,1000,1000)

plt.plot(tt,xx)
plt.axhline(0, linestyle='--' , color = 'black', linewidth=1)
```

En este plot podemos ver la función llamando con una frecuencia de 2Hz

Ahora veremos las experiencias realizadas en clase con freciencias de 500Hz, 999Hz, 1001Hz y 2001Hz

```python
tt , xx = generador(1,0,500,0,1000,1000)

plt.plot(tt,xx)
plt.axhline(0, linestyle='--' , color = 'black', linewidth=1)
```

```python
tt , xx = generador(1,0,999,0,1000,1000)

plt.plot(tt,xx)
plt.axhline(0, linestyle='--' , color = 'black', linewidth=1)
```

```python
tt , xx = generador(1,0,1001,0,1000,1000)

plt.plot(tt,xx)
plt.axhline(0, linestyle='--' , color = 'black', linewidth=1)
```

```python
tt , xx = generador(1,0,2001,0,1000,1000)

plt.plot(tt,xx)
plt.axhline(0, linestyle='--' , color = 'black', linewidth=1)
```
